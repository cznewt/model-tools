# 02 Helm: text templating

`values.yaml` is the chart API, `values-<env>.yaml` refines it, `values.schema.json` is the only type check. Secrets come in through `vals` at render time.


```bash
cd /source/work/gitops-renderers
cat examples/02-helm/web/values.yaml; echo ---; cat examples/02-helm/values-prod.yaml
```

```bash
cd /source/work/gitops-renderers
helm dependency update examples/02-helm/web >/dev/null && helm lint examples/02-helm/web -f examples/02-helm/values-prod.yaml
```

```bash
cd /source/work/gitops-renderers
export SOPS_AGE_KEY_FILE=$PWD/examples/secrets/demo.agekey
vals eval -f examples/secrets/values-prod-secrets.yaml
```

```bash
cd /source/work/gitops-renderers
helm template web examples/02-helm/web -n web --skip-tests -f examples/02-helm/values-prod.yaml | yq 'select(.kind == "Deployment" and .metadata.name == "web") | .spec.template.spec.containers[0]'
```

Try it: set `replicaCount: "three"` in values-prod.yaml and run `helm lint` again. Where does the error surface, and how readable is it?
