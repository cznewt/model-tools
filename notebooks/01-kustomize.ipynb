# 01 Kustomize: data plus patches

The base is the plain-YAML reference for every other renderer. The overlay *is* the context: every environment-specific value is a patch.


```bash
cd /source/work/gitops-renderers
cat examples/01-kustomize/base/kustomization.yaml; echo; cat examples/01-kustomize/overlays/prod/kustomization.yaml
```

```bash
cd /source/work/gitops-renderers
kustomize build --enable-helm examples/01-kustomize/overlays/prod > /tmp/prod.yaml && grep -c '^kind:' /tmp/prod.yaml && yq 'select(.kind == "Deployment" and .metadata.name == "web") | .spec.replicas, .spec.template.spec.containers[0].resources' /tmp/prod.yaml
```

```bash
cd /source/work/gitops-renderers
diff <(kustomize build --enable-helm examples/01-kustomize/overlays/dev | yq 'select(.kind == "Ingress")') <(yq 'select(.kind == "Ingress")' /tmp/prod.yaml) || true
```

Try it: change `namePrefix` in the prod overlay and rebuild. What else would you have to touch for a third environment?
