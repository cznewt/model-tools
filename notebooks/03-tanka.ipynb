# 03 Jsonnet with Tanka

`lib/web.libsonnet` is the workload as a function of its context; `lib/contexts.libsonnet` is the cascade; `spec.json` is what Tanka itself knows about an environment.

```bash
cd /source/work/gitops-renderers/examples/03-tanka
export HOME=/tmp
jb install >/dev/null && tk tool charts vendor >/dev/null && tk eval environments/web/prod -e '$._config' 2>/dev/null || tk eval environments/web/prod | head -40
```
