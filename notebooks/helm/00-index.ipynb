## Exercises

The prompts left in the notebooks, in one place.

**[Helm 01: a chart is a directory](01-chart-anatomy.ipynb)**

- compare the scaffold's deployment template with the companion chart's and list what you would delete.

**[Helm 02: values are the context](02-values-and-schema.ipynb)**

- mark a field required in values.schema.json, render without it, and read the error.

**[Helm 03: dependencies, repositories and OCI](03-dependencies-and-oci.ipynb)**

- pin podinfo one minor version lower, update the lock, and see what changes in the output.

**[Helm 04: testing a chart and handling secrets](04-testing-and-secrets.ipynb)**

- point the vals reference at a key that does not exist and find where the render fails.
