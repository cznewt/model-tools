# Helm 01: a chart is a directory

`helm create` scaffolds the canonical layout: `Chart.yaml`, `values.yaml`, `templates/` with `_helpers.tpl`, `NOTES.txt` and tests. Helm 4 keeps this model.


```bash
export HOME=/tmp
helm version --short
mkdir -p /source/work && cd /source/work && rm -rf helm-lab && mkdir helm-lab && cd helm-lab && helm create scaffold >/dev/null && find scaffold -type f | sort
```

```bash
cd /source/work/helm-lab
helm lint scaffold && helm template demo scaffold | grep -E '^kind:' | sort | uniq -c
```

```bash
cd /source/work/helm-lab
sed -n 1,40p scaffold/templates/deployment.yaml
```

The companion chart is smaller and adds what the scaffold leaves out: a `values.schema.json`, a subchart dependency and per-environment values files outside the chart.


```bash
export HOME=/tmp
mkdir -p /source/work && cd /source/work && [ -d gitops-renderers ] || git clone -q --recurse-submodules https://github.com/cznewt/gitops-renderers.git
cd /source/work/gitops-renderers/examples/02-helm && find . -type f | sort && cat web/Chart.yaml
```

```bash
cd /source/work/gitops-renderers/examples/02-helm
helm show values web | head -30
```
