# Jsonnet with Tanka

A project from tk init: environments, jsonnet-bundler, charts and Kustomize inside Tanka, inline environments and tests.

Every cell is a shell command, so the same steps work in a terminal. Run the notebooks in order; `/source/work` survives restarts.
