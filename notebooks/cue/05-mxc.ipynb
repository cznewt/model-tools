# CUE 05: mxc, modelling the fleet

[epcim/mxc](https://github.com/epcim/mxc) sits one level above the other renderers: a typed model of `#Topology` -> `#Location` -> `#Cluster` -> `#App` from which adapters project Helm values, Kustomize overlays, Argo CD applications, Kluctl deployments, Compose files or Terraform variables. It is vendored in the companion repository as a pinned submodule.


```bash
export HOME=/tmp
mkdir -p /source/work && cd /source/work && [ -d gitops-renderers ] || git clone -q --recurse-submodules https://github.com/cznewt/gitops-renderers.git
cd /source/work/gitops-renderers/examples/06-mxc && cat README.md | head -30 && git -C mxc log --oneline -1
```

```bash
cd /source/work/gitops-renderers/examples/06-mxc/mxc
ls module/schema | head -20; echo; grep -n '^#AppMxc\|^#ClusterMxc\|^#Topology\|^#Location\|^#WithNetwork' module/schema/*.cue | head -10
```

```bash
cd /source/work/gitops-renderers/examples/06-mxc/mxc
sed -n 1,45p examples/cluster-standalone/globals.cue
```

```bash
cd /source/work/gitops-renderers/examples/06-mxc/mxc
sed -n 1,40p examples/cluster-standalone/apps.cue
```

Read `apps.cue` as a context: the app declares intent (chart, image, ports, expose) against `schema.#AppMxc`, and the platform facts in `globals.cue` (domain, storage class, ingress annotations) cascade into every projection. The upstream `just` recipes validate and export; at the pinned commit the export does not evaluate under CUE 0.17, which is exactly the kind of drift a pinned dependency is for.


```bash
cd /source/work/gitops-renderers/examples/06-mxc/mxc
export HOME=/tmp
(just mxc::vet examples/cluster-standalone >/dev/null 2>&1 && echo 'vet: ok') || echo 'vet: fails at the pinned commit (see README)'
(cd module && cue vet ./schema/... >/dev/null 2>&1 && echo 'schema vet: ok') || echo 'schema vet: needs the registry modules (cue.dev) resolved first'
```
