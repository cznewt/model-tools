# CUE 03: contexts and rendering in the companion repository

`examples/05-cue` describes the talk's workload: `schema.cue` is the context schema, `contexts.cue` the environments, `resources.cue` the workload as a function of the context, and `render_tool.cue` a `cue cmd` that writes YAML streams.


```bash
export HOME=/tmp
mkdir -p /source/work && cd /source/work && [ -d gitops-renderers ] || git clone -q --recurse-submodules https://github.com/cznewt/gitops-renderers.git
cd /source/work/gitops-renderers/examples/05-cue && ls web && cue vet ./web && echo vet ok
```

```bash
cd /source/work/gitops-renderers/examples/05-cue
cue eval ./web -e contexts.prod
```

```bash
cd /source/work/gitops-renderers/examples/05-cue
cue export ./web -e envs.dev.resources.ingress --out yaml
```

Refinement, not override: give prod an ingress class outside the allowed set and CUE points at the disjunction.


```bash
cd /source/work/gitops-renderers/examples/05-cue
sed -i 's/ingressClass: "nginx"/ingressClass: "haproxy"/' web/contexts.cue && (cue vet ./web 2>&1 | head -5) || true; git checkout web/contexts.cue
```

```bash
cd /source/work/gitops-renderers/examples/05-cue
cat web/render_tool.cue && cue cmd render ./web && git status --short ../../rendered/cue && echo "(no diff: rendering is deterministic)"
```
