# Kapitan 01: the reference setup

[kapicorp/kapitan-reference](https://github.com/kapicorp/kapitan-reference) is the official
bootstrap repository: an inventory of targets and classes, the Kubernetes and Terraform
generators, jinja2 scripts and docs, refs for secrets, and every compiled target committed under
`compiled/`. Upstream drives it through `./kapitan`, a wrapper around the kapitan Docker image;
here the binary is already in the image, so the wrapper is not needed.


```bash
export HOME=/tmp
mkdir -p /source/work && cd /source/work && [ -d kapitan-reference ] || git clone -q --depth 1 https://github.com/kapicorp/kapitan-reference.git
cd kapitan-reference && git log --oneline -1 && kapitan --version && ls
```

```bash
cd /source/work/kapitan-reference
cat .kapitan
```

`.kapitan` holds the CLI defaults: prune, embed refs into the compiled output, fetch missing dependencies, where refs live, search paths for Jsonnet and the generators.


```bash
cd /source/work/kapitan-reference
kapitan inventory --targets 2>/dev/null | head -25 || find inventory/targets -name '*.yml' | sort
```

```bash
cd /source/work/kapitan-reference
cat inventory/targets/examples/tutorial.yml; echo ---; cat inventory/classes/common.yml
```

```bash
cd /source/work/kapitan-reference
kapitan compile -t tutorial 2>&1 | tail -3 && find compiled/tutorial -type f | sort
```

The first compile fetched `kapicorp/generators` into `system/lib` (the `kgenlib` class declares it as a git dependency). One target produced manifests, scripts and a README from the same inventory.


```bash
cd /source/work/kapitan-reference
head -20 compiled/tutorial/README.md; echo ...; yq 'select(.kind == "Deployment") | .spec.template.spec.containers[0] | .image, .ports' compiled/tutorial/manifests/echo-server-bundle.yml
```

Try it: compile the vault target and compare its manifests with the tutorial's.
