# Kapitan 05: jinja2 scripts, docs and Helm charts

Kapitan compiles more than manifests. The same inventory feeds jinja2 templates (per-target
`kubectl` wrappers, apply scripts, a README with component tables) and the `helm` input type,
which renders vendored charts with values from the inventory.


```bash
cd /source/work/kapitan-reference
export HOME=/tmp
kapitan inventory -t tutorial -p parameters.kapitan.compile 2>/dev/null | head -40
```

```bash
cd /source/work/kapitan-reference
ls system/templates/scripts/includes system/templates/scripts/kapitan; echo; sed -n 1,30p system/templates/scripts/includes/kapitan.include
```

```bash
cd /source/work/kapitan-reference
ls compiled/tutorial/scripts; echo; sed -n 1,30p compiled/tutorial/scripts/apply
```

The rendered `scripts/kubectl` and `scripts/apply` carry the target's context and namespace, so the CI of the reference repository can create a kind cluster and apply the tutorial target with two calls. Docs come from the same mechanism.


```bash
cd /source/work/kapitan-reference
ls system/templates/docs; echo; sed -n 1,40p system/templates/docs/README.md
```

```bash
cd /source/work/kapitan-reference
sed -n 1,45p inventory/classes/components/helm/cert-manager-helm.yml
```

A Helm chart is a dependency (`type: helm`, fetched into `system/sources/charts`) plus a `helm` compile entry with `helm_values` from the inventory: the chart's `values.yaml` becomes one more layer of the model.


```bash
cd /source/work/kapitan-reference
grep -rln 'components.helm.cert-manager' inventory/targets inventory/classes | head -3; sed -n 45,80p inventory/classes/components/helm/cert-manager-helm.yml
```
