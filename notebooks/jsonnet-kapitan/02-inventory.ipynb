# Kapitan 02: the inventory is the model

reclass composes classes into targets: `common` -> `kapitan.common` -> generator classes, then
component and application classes, then the target's own parameters. `${...}` interpolation
crosses class boundaries, which is how a component derives values from the target it lands in.


```bash
cd /source/work/kapitan-reference
export HOME=/tmp
cat inventory/classes/kapitan/common.yml | head -40
```

```bash
cd /source/work/kapitan-reference
kapitan inventory -t mysql -p parameters.components.mysql 2>/dev/null | head -40
```

`kapitan inventory -t <target> -p <path>` shows the fully resolved value: `${target_name}` became `mysql`, the refs got their per-target path.


```bash
cd /source/work/kapitan-reference
cat inventory/targets/kapicorp/dev-sockshop.yml; echo ---; cat inventory/targets/kapicorp/prod-sockshop.yml
```

```bash
cd /source/work/kapitan-reference
diff <(kapitan inventory -t dev-sockshop -p parameters.components.carts 2>/dev/null) <(kapitan inventory -t prod-sockshop -p parameters.components.carts 2>/dev/null) || true
```

Two environments of the sock-shop application differ by a handful of lines in the target files; the `applications.sock-shop.component_defaults` cascade in `inventory/classes/applications/sock-shop.yml` sets what every component shares.
