## Exercises

The prompts left in the notebooks, in one place.

**[Kapitan 01: the reference setup](01-reference-setup.ipynb)**

- compile the vault target and compare its manifests with the tutorial's.

**[Kapitan 02: the inventory is the model](02-inventory.ipynb)**

- add a class that sets one label on every component and find which targets change.

**[Kapitan 03: the Kubernetes generator](03-generators.ipynb)**

- give the hello component a config map rendered from a template, then recompile.

**[Kapitan 04: refs, the secrets story](04-refs-and-secrets.ipynb)**

- write a second plain ref, reveal it, then delete the ref file and read what compile says.

**[Kapitan 05: jinja2 scripts, docs and Helm charts](05-templates-scripts-docs.ipynb)**

- add a jinja2 template that prints the component table for a target.

**[Kapitan 06: validation and testing](06-validate-and-test.ipynb)**

- break one class with a typo and read what kapitan lint reports.
