## Exercises

The prompts left in the notebooks, in one place.

**[01 Kustomize: data plus patches](01-kustomize.ipynb)**

- change `namePrefix` in the prod overlay and rebuild. What else would you have to touch for a third environment?

**[02 Helm: text templating](02-helm.ipynb)**

- set `replicaCount: "three"` in values-prod.yaml and run `helm lint` again. Where does the error surface, and how readable is it?

**[03 Jsonnet with Tanka](03-tanka.ipynb)**

- add `replicas: 5` to `contexts.prod` and run `tk show` again, then break it with `replicas: '5'` (a string). Jsonnet renders both.

**[04 Jsonnet with Kapitan](04-kapitan.ipynb)**

- add a `cluster.staging` class and a `staging` target that composes it. How many lines did the new environment cost?

**[05 CUE and Timoni](05-cue.ipynb)**

- give `ingress.className` a value outside the allowed set in `values-prod.cue` and build again.

**[06 Secrets: reference, encrypt, generate](06-secrets.ipynb)**

- which of the three patterns keeps the rendered manifests reviewable? Which one lets an AI agent see a real secret if you are careless?
