# 04 Jsonnet with Kapitan

The reclass inventory is the model of the fleet: `common` -> `cluster.*` -> `component.*` composed by targets, with `${interpolation}` across classes. Refs keep secrets out of the compiled output until `--reveal`.


```bash
cd /source/work/gitops-renderers/examples/04-kapitan
cat inventory/targets/prod.yml; echo ---; cat inventory/classes/component/web.yml
```

```bash
cd /source/work/gitops-renderers/examples/04-kapitan
kapitan inventory -t prod 2>/dev/null | yq '.parameters.web'
```

```bash
cd /source/work/gitops-renderers/examples/04-kapitan
kapitan compile --fetch --use-go-jsonnet --search-paths . lib --refs-path refs --output-path . -t prod 2>&1 | tail -2 && grep -A1 API_TOKEN compiled/prod/manifests/secret.yml
```

```bash
cd /source/work/gitops-renderers/examples/04-kapitan
kapitan compile --reveal --use-go-jsonnet --search-paths . lib --refs-path refs --output-path . -t prod 2>&1 | tail -1 && grep -A1 API_TOKEN compiled/prod/manifests/secret.yml
```
