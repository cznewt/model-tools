# 05 CUE and Timoni

Types are values. `#Context` carries constraints and defaults, `contexts.cue` refines it, `cue vet` tells you where you are wrong with a path. Timoni packages the same idea with a lifecycle.


```bash
cd /source/work/gitops-renderers/examples/05-cue
cue vet ./web && cue eval ./web -e contexts.prod
```

```bash
cd /source/work/gitops-renderers/examples/05-cue
cue export ./web -e envs.prod.resources.deployment --out yaml
```

```bash
cd /source/work/gitops-renderers/examples/05-cue
sed -i 's/replicas: 3/replicas: 30/' web/contexts.cue && (cue vet ./web || true); git checkout web/contexts.cue
```

```bash
cd /source/work/gitops-renderers
export HOME=/tmp
timoni build web examples/05-cue/timoni/web -n web -f examples/05-cue/timoni/values-prod.cue | yq 'select(.kind == "Deployment") | .spec.replicas, .metadata.labels'
```

Try it: give `ingress.className` a value outside the allowed set in `values-prod.cue` and build again.
