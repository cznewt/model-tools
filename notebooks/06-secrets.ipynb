# 06 Secrets: reference, encrypt, generate

Three patterns side by side: an `ExternalSecret` reference in every renderer, a SOPS plus age encrypted value consumed by Kustomize and Helm (via vals), and Kapitan refs generated at compile time.


```bash
cd /source/work/gitops-renderers
export SOPS_AGE_KEY_FILE=$PWD/examples/secrets/demo.agekey
head -12 examples/secrets/web-secrets.enc.yaml; echo ...; sops --decrypt examples/secrets/web-secrets.enc.yaml | yq '.stringData'
```

```bash
cd /source/work/gitops-renderers
grep -h -A3 'kind: ExternalSecret' rendered/*/prod.yaml | grep -c ExternalSecret; yq 'select(.kind == "ExternalSecret") | .spec' rendered/cue/prod.yaml
```
